I learned how to Push my Code from VS code to Github.

Steps :

1. Install Git to system.
2. To Know git is installed Open terminal >- git --version
3. To SetUp name and email >- git config --global user.name "Your Name"
                              git config --global user.email "your.email@example.com"
4. Initialize Git in a Project >- git init
5. Add Files to Staging >- git add filename.txt or git add . (Add All Files)
6. Commit Changes >-  git commit -m "Initial commit"
7. Connect to GitHub >- git remote add origin https://github.com/username/repository.git
8. Push Your Code >- git push -u origin main
9. Check Status Anytime >- git status
